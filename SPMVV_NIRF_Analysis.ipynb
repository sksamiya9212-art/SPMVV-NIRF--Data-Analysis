```python
placement_df.columns = placement_df.columns.str.strip()
placement_df.columns = placement_df.columns.str.lower()
```

```python
placement_df = placement_df[['department', 'name of the company', 'package']]
```

```python
placement_df['package'] = placement_df['package'].str.replace(' LPA', '')
placement_df['package'] = placement_df['package'].astype(float)
```

```python
placement_df.loc[:, 'package'] = placement_df['package']
```

```python
import numpy as np

def clean_package(value):
    if isinstance(value, str):
        value = value.replace(' LPA', '').strip()

        if '-' in value:
            try:
                parts = value.split('-')
                return (float(parts[0]) + float(parts[1])) / 2
            except:
                return np.nan

        else:
            try:
                return float(value)
            except:
                return np.nan

    return value

placement_df['package'] = placement_df['package'].apply(clean_package)
```

```python
placement_df = placement_df.dropna(subset=['package'])
```

```python
dept_counts = placement_df['department'].value_counts()
print(dept_counts)
```

```python
import matplotlib.pyplot as plt

dept_counts.plot(kind='bar')
plt.title("Placements by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.show()
```

```python
avg_salary.plot(kind='bar')
plt.title("Average Salary by Department")
plt.ylabel("Package (LPA)")
plt.show()
```

```python
patent_df['Patent Filed Date (DD/MM/YYYY)'] = pd.to_datetime(
    patent_df['Patent Filed Date (DD/MM/YYYY)'],
    errors='coerce'
)
```

```python
patent_df['year'] = patent_df['Patent Filed Date (DD/MM/YYYY)'].dt.year
```

```python
patents_per_year = patent_df.groupby('year').size().reset_index(name='patent_count')
patents_per_year
```

```python
import matplotlib.pyplot as plt

patents_per_year.plot(x='year', y='patent_count', kind='line', marker='o')
plt.title("Patents Trend Over Years")
plt.xlabel("Year")
plt.ylabel("Number of Patents")
plt.show()
```
